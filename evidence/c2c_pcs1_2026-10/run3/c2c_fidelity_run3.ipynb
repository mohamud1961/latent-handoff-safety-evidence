# PCS x C2C, run 3: a strong sharer (Qwen3-4B) into a weak receiver (Qwen3-0.6B). Does the bridge communicate *content*?

**Question.** Cache-to-Cache (C2C, ICLR 2026, arXiv 2510.03215) fuses a frozen *sharer* LLM's KV cache into a frozen *receiver*'s. Papers score task accuracy. PCS asks whether the gain depends on the sharer's cache *for this question* (several wrong-question controls), survives a label-free calibration of the receiver's letter bias, shows up on a **non-letter task** (GSM8K exact-answer generation), and whether a premise only the sharer sees reaches the receiver (planted belief).

Runs 1 and 2 (Qwen2.5-0.5B -> Qwen3-0.6B, Qwen2.5-1.5B -> Qwen3-1.7B) found the accuracy change was reproduced by a wrong-question sharer cache and by calibration. **Run 3** uses the released general fuser `nics-efc/C2C_Fuser` / `qwen3_0.6b+qwen3_4b_Fuser/final`: receiver `Qwen/Qwen3-0.6B`, sharer `Qwen/Qwen3-4B` (the paper's Table 9 reports this pair: OpenBookQA R 39.2 / S 79.4 / C2C 55.2). The sharer is far stronger than the receiver, so there is a lot to transmit. The pair is a parameter (`PAIRS`).

**Read `PREREG_run3.md` first** (and `PREREG.md`, `PREREG_run2.md`). They fix hypotheses, sample sizes, metrics and decision rules before any result exists.

| Part | What | Items | Approx. T4 time |
|---|---|---|---|
| A | plumbing vs the paper's OpenBookQA numbers; generation readout, logit readout, mismatched sharer under generation | 200 | 15-20 min |
| B | **MC fidelity**: C vs K=3 wrong-question controls C_mm,k on every item, calibrated receiver, inheritance, text handoffs | 2000 | 45-60 min |
| C | planted false belief only the sharer sees (exploratory but kept) | 500 | 8-10 min |
| N | **numeric task**: GSM8K exact-answer generation through C2C's own prompt and `GSM8KEvaluator`; R, S, C, C_mm; numeric inheritance and leak check | ~150-300 | up to 50 min (stops at a deadline) |

Setup (installs, ~8 GB of model downloads, an fp32 CPU reference for the fp16 gate) takes 20-30 min. Everything is greedy and seeded. Outputs: `results.json`, written to `/kaggle/working` on Kaggle or the current directory elsewhere.

**Run it:** Kaggle: Settings > Accelerator > **GPU T4**, Internet **on**, Run All. Rehearsal with small sizes (not for results): env `PCS_QUICK=1`.

```python
# --- 6. Models + fuser (dtype-aware loader; the repo's own loader hard-codes bf16, which a T4 cannot do natively) ---
def fuser_dir():
    pats = [f"{FUSER_SUB}/final/*.json"] if SMOKE else [f"{FUSER_SUB}/final/*", f"{FUSER_SUB}/config.json"]
    root = snapshot_download(repo_id=FUSER_REPO, allow_patterns=pats)
    return os.path.join(root, FUSER_SUB, "final")

FUSER_PATH = fuser_dir()
print("fuser dir:", FUSER_PATH, "| files:", len(os.listdir(FUSER_PATH)))

def _tiny(cfg_id, **over):
    cfg = AutoConfig.from_pretrained(cfg_id)
    for k, v in over.items():
        setattr(cfg, k, v)
    return cfg

def build_stack(dtype, device=None):
    """Receiver + sharer + fuser in `dtype` on `device` (default: the global DEVICE)."""
    device = device if device is not None else DEVICE
    tok_r = AutoTokenizer.from_pretrained(RECV_ID); set_default_chat_template(tok_r, RECV_ID)
    tok_s = AutoTokenizer.from_pretrained(SHAR_ID)
    if SMOKE:
        torch.manual_seed(SEED)
        recv = AutoModelForCausalLM.from_config(_tiny(RECV_ID, **PAIR["smoke_recv"]), torch_dtype=dtype)
        shar = AutoModelForCausalLM.from_config(_tiny(SHAR_ID, **PAIR["smoke_shar"]), torch_dtype=dtype)
    else:
        recv = AutoModelForCausalLM.from_pretrained(RECV_ID, torch_dtype=dtype)
        shar = AutoModelForCausalLM.from_pretrained(SHAR_ID, torch_dtype=dtype)
    recv, shar = recv.to(device).eval(), shar.to(device).eval()
    n_proj = len([f for f in os.listdir(FUSER_PATH) if re.fullmatch(r"projector_\d+\.json", f)])
    assert n_proj == recv.config.num_hidden_layers, f"fuser has {n_proj} projectors, receiver has {recv.config.num_hidden_layers} layers"
    projs = []
    for t_ in range(n_proj):
        over = {"dtype": dtype}
        if SMOKE:
            over.update(hidden_dim=64, intermediate_dim=64)
        p = load_projector(os.path.join(FUSER_PATH, f"projector_{t_}.json"), override_args=over)
        pt = os.path.join(FUSER_PATH, f"projector_{t_}.pt")
        if not SMOKE:
            res = p.load_state_dict(torch.load(pt, map_location="cpu", weights_only=True), strict=False)   # repo uses strict=False; we check
            if res.missing_keys or res.unexpected_keys:
                raise RuntimeError(f"fuser projector_{t_}: missing={res.missing_keys} unexpected={res.unexpected_keys}")
            bad = [k for k, v in p.state_dict().items() if v.is_floating_point() and not torch.isfinite(v).all()]
            if bad:   # e.g. bf16 -> fp16 overflow while casting the released weights
                raise RuntimeError(f"fuser projector_{t_}: non-finite weights after cast to {dtype}: {bad[:5]}")
        if SMOKE:   # random projectors start with closed gates (logit 0 -> no fusion); open them so the fusion path is exercised
            with torch.no_grad():
                p.key_gate_logit.fill_(1.0); p.value_gate_logit.fill_(1.0)
        projs.append(p.to(device))
    ros = RosettaModel(model_list=[recv, shar], base_model_idx=0, projector_list=projs).to(device).eval()
    ros.load_projector_config(os.path.join(FUSER_PATH, "projector_config.json"))
    assert len(ros.projector_dict[0][1]) == recv.config.num_hidden_layers, "projector map must cover every receiver layer"
    return dict(tok_r=tok_r, tok_s=tok_s, recv=recv, shar=shar, ros=ros, dtype=dtype, n_proj=n_proj)

def use_stack(st):
    global TOK_R, TOK_S, RECV, SHAR, ROS, DTYPE_USED, OPT_IDS, OPT_T, RESP_IDS, EOS_R, EOS_S, PAD_R, PAD_S
    TOK_R, TOK_S, RECV, SHAR, ROS, DTYPE_USED = st["tok_r"], st["tok_s"], st["recv"], st["shar"], st["ros"], st["dtype"]
    o_r, o_s = get_option_token_ids(TOK_R, 4), get_option_token_ids(TOK_S, 4)
    assert o_r == o_s, "receiver/sharer tokenizers disagree on option-letter ids"
    OPT_IDS = o_r
    OPT_T = torch.tensor(OPT_IDS, device=DEVICE)
    r1, r2 = TOK_R(RESP_TEXT, add_special_tokens=False)["input_ids"], TOK_S(RESP_TEXT, add_special_tokens=False)["input_ids"]
    assert r1 == r2, "response-suffix tokenization differs between tokenizers"
    RESP_IDS = r1
    EOS_R, EOS_S = RECV.generation_config.eos_token_id, SHAR.generation_config.eos_token_id
    PAD_R = TOK_R.pad_token_id if TOK_R.pad_token_id is not None else (EOS_R[0] if isinstance(EOS_R, list) else EOS_R)
    PAD_S = TOK_S.pad_token_id if TOK_S.pad_token_id is not None else (EOS_S[0] if isinstance(EOS_S, list) else EOS_S)

def free_stack(st):
    for k in list(st):
        st[k] = None
    for n_ in ("RECV", "SHAR", "ROS"):
        globals()[n_] = None
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

```python
# --- 7. Inference primitives (all conditions share one readout: letter logits after "The correct answer is") ---
COUNTERS = {"nonfinite": 0}

def encode_user(tok, user_text):
    s = tok.apply_chat_template([{"role": "user", "content": user_text}], tokenize=False,
                                add_generation_prompt=True, enable_thinking=False)
    return tok(s, add_special_tokens=False)["input_ids"]

def to_pred(logits_vec):
    """a = argmax letter (ties -> lower letter); p = softmax over the four letter logits; lp = log-softmax over them (the
    calibrated baselines of PREREG_run2 (a), kept in run 3 use lp)."""
    sel = logits_vec[OPT_T].float()
    if not torch.isfinite(sel).all():
        COUNTERS["nonfinite"] += 1
        return {"a": "?", "p": [None] * 4, "lp": [None] * 4}
    lp = torch.log_softmax(sel, 0).cpu().numpy()
    p = np.exp(lp)
    return {"a": LETTERS[int(lp.argmax())], "p": [round(float(x), 4) for x in p], "lp": [round(float(x), 5) for x in lp]}

@torch.no_grad()
def lm_logits(model, ids, mask=None, positions=None):
    """Plain HF forward (receiver-alone / sharer-alone). Returns last-position logits."""
    x = torch.tensor([ids], device=DEVICE)
    pos = torch.arange(x.shape[1], device=DEVICE).unsqueeze(0) if positions is None else torch.tensor([positions], device=DEVICE)
    am = None if mask is None else torch.tensor([mask], device=DEVICE)
    return model(input_ids=x, attention_mask=am, position_ids=pos, use_cache=False, logits_to_keep=1).logits[0, -1]

def _kv_index(n_instr, n_total, sharer_mask=1):
    a = torch.tensor([sharer_mask, 0], dtype=torch.long).repeat(n_instr, 1).unsqueeze(0).to(DEVICE)
    b = torch.tensor([-1, 0], dtype=torch.long).repeat(n_total - n_instr, 1).unsqueeze(0).to(DEVICE)
    return [a, b]

@torch.no_grad()
def c2c_logits(ids_r, ids_s, n_instr, mask_r=None, mask_s=None, sharer_mask=1, as_tensor=False):
    """C2C-fused forward. ids_r / ids_s: receiver / sharer token ids, same length. The first n_instr positions are
    fused (kv_cache_index [1,0]); the trailing response tokens are not (and the sharer never sees them: C2C only
    computes sharer caches for the fused sections)."""
    assert len(ids_r) == len(ids_s), "C2C needs position-aligned inputs (equal length)"
    n = len(ids_r)
    xr, xs = torch.tensor([ids_r], device=DEVICE), torch.tensor([ids_s], device=DEVICE)
    mr = torch.tensor([mask_r if mask_r is not None else [1] * n], device=DEVICE)
    ms = torch.tensor([mask_s if mask_s is not None else [1] * n], device=DEVICE)
    pos = torch.arange(n, device=DEVICE).unsqueeze(0)
    kv = _kv_index(n_instr, n, sharer_mask)
    if as_tensor:   # exactly the call pattern of C2C's own evaluator for same-tokenizer pairs
        out = ROS.forward(kv_cache_index=kv, input_ids=xr, attention_mask=mr, position_ids=pos)
    else:
        out = ROS.forward(kv_cache_index=kv, input_ids=[xr, xs], attention_mask=[mr, ms], position_ids=pos, use_cache=True)
    return out.logits[0, -1]

def _gen_cfg(n_new, eos, pad):
    return GenerationConfig(do_sample=False, max_new_tokens=n_new, pad_token_id=pad, eos_token_id=eos)

@torch.no_grad()
def hf_generate(model, tok, ids, n_new, eos, pad):
    x = torch.tensor([ids], device=DEVICE)
    out = model.generate(input_ids=x, attention_mask=torch.ones_like(x), generation_config=_gen_cfg(n_new, eos, pad))
    return tok.decode(out[0, x.shape[1]:], skip_special_tokens=True).strip("\n")

@torch.no_grad()
def c2c_generate(ids, n_new=64, ids_s=None):
    """README / evaluator pattern: fuse all prompt tokens except the last (response_length=1), greedy decode.
    ids_s (optional): sharer tokens, same length as ids. Used for the mismatched-sharer control under the generation
    readout (PREREG_run2 (c), kept in run 3): the sharer cache is computed on a different same-length prompt."""
    x = torch.tensor([ids], device=DEVICE)
    n = x.shape[1]
    pos = torch.arange(n, device=DEVICE).unsqueeze(0)
    if ids_s is None:
        out = ROS.generate(kv_cache_index=_kv_index(n - 1, n), input_ids=x, attention_mask=torch.ones_like(x),
                           position_ids=pos, do_sample=False, max_new_tokens=n_new)
    else:
        assert len(ids_s) == n, "C2C needs position-aligned inputs (equal length)"
        xs = torch.tensor([ids_s], device=DEVICE)
        out = ROS.generate(kv_cache_index=_kv_index(n - 1, n), input_ids=[x, xs], attention_mask=[torch.ones_like(x), torch.ones_like(xs)],
                           position_ids=pos, do_sample=False, max_new_tokens=n_new)
    return TOK_R.decode(out[0, n:], skip_special_tokens=True).strip("\n")

@torch.no_grad()
def hf_generate_batch(model, tok, id_lists, n_new, eos, pad, bs=16):
    outs = []
    for i in range(0, len(id_lists), bs):
        chunk = id_lists[i:i + bs]
        m = max(len(c) for c in chunk)
        x = torch.tensor([[pad] * (m - len(c)) + c for c in chunk], device=DEVICE)
        am = torch.tensor([[0] * (m - len(c)) + [1] * len(c) for c in chunk], device=DEVICE)
        out = model.generate(input_ids=x, attention_mask=am, generation_config=_gen_cfg(n_new, eos, pad))
        outs += [tok.decode(o[m:], skip_special_tokens=True) for o in out]
    return outs
```

```python
# --- 8. Statistics helpers (bootstrap over items) ---
def boot_ci(x, n_boot=None, seed=0):
    """Percentile bootstrap CI of the mean. Pass per-item differences for paired comparisons."""
    x = np.asarray(x, dtype=float)
    n = len(x)
    if n == 0:
        return (float("nan"),) * 3
    n_boot = n_boot or CFG["N_BOOT"]
    rng = np.random.default_rng(seed)
    chunk = max(1, int(2_000_000 // n))
    means, done = [], 0
    while done < n_boot:
        m = min(chunk, n_boot - done)
        means.append(x[rng.integers(0, n, size=(m, n))].mean(1))
        done += m
    means = np.concatenate(means)
    return float(x.mean()), float(np.percentile(means, 2.5)), float(np.percentile(means, 97.5))

def boot_ci_unpaired_diff(xa, xb, n_boot=None, seed=0):
    xa, xb = np.asarray(xa, float), np.asarray(xb, float)
    if len(xa) == 0 or len(xb) == 0:
        return (float("nan"),) * 3
    n_boot = n_boot or CFG["N_BOOT"]
    rng = np.random.default_rng(seed)
    da = np.array([xa[rng.integers(0, len(xa), len(xa))].mean() for _ in range(n_boot)])
    db = np.array([xb[rng.integers(0, len(xb), len(xb))].mean() for _ in range(n_boot)])
    d = da - db
    return float(xa.mean() - xb.mean()), float(np.percentile(d, 2.5)), float(np.percentile(d, 97.5))

def fmt(ci, pct=False):
    m, lo, hi = ci
    if m != m:
        return "n/a"
    return f"{m*100:5.1f}% [{lo*100:5.1f},{hi*100:5.1f}]" if pct else f"{m:+.3f} [{lo:+.3f},{hi:+.3f}]"

def table(rows, header):
    if not rows:
        print(*header); return
    w = [max(len(str(h)), *(len(str(r[i])) for r in rows)) for i, h in enumerate(header)]
    line = "  ".join(str(h).ljust(w[i]) for i, h in enumerate(header))
    print(line); print("-" * len(line))
    for r in rows:
        print("  ".join(str(c).ljust(w[i]) for i, c in enumerate(r)))
```

```python
# --- 9. Build items: sample, tokenize, length-filter ---
def seeded_sample(rows, n, salt):
    rows = sorted(rows, key=lambda r: r["uid"])
    random.Random(f"{SEED}:{salt}").shuffle(rows)
    return rows[:n], rows[n:]

def finish_item(r):
    r = dict(r)
    r["id"] = f"{r['bench']}:{r['uid']}"
    r["prompt"] = make_prompt(r["question"], r["choices"])
    r["ids"] = encode_user(TOK_R, r["prompt"])           # receiver-template tokens (what the fuser feeds BOTH models)
    r["ids_nat"] = encode_user(TOK_S, r["prompt"])       # sharer's own chat template (sensitivity only)
    r["L"] = len(r["ids"])
    return r

def build_items():
    obqa, _ = seeded_sample(POOLS["obqa"], CFG["N_OBQA"], "obqa")
    arc, _ = seeded_sample(POOLS["arc"], CFG["N_ARC"], "arc")
    mmlu, mmlu_rest = seeded_sample(POOLS["mmlu"], CFG["N_MMLU"], "mmlu")
    reserve = mmlu_rest[:CFG["N_RESERVE_MMLU"]]
    pool = [finish_item(r) for r in mmlu + arc + obqa]
    random.Random(f"{SEED}:order").shuffle(pool)          # pre-registered processing order (stage-wise chunks are unbiased)
    reserve = [finish_item(r) for r in reserve]
    n0 = len(pool)
    pool = [it for it in pool if it["L"] <= CFG["MAX_TOKENS"]]
    reserve = [it for it in reserve if it["L"] <= CFG["MAX_TOKENS"]]
    if len(pool) < n0:
        DEVIATIONS.append(f"dropped {n0-len(pool)} items longer than {CFG['MAX_TOKENS']} tokens (PREREG item filter)")
    return pool, reserve
```

```python
# --- 10. Load models (dtype policy + fp16 sanity gate), build items, partner bank, early timing estimate ---
import contextlib

@contextlib.contextmanager
def on_device(dev):
    """Temporarily point the inference primitives (which read the global DEVICE) at another device."""
    global DEVICE
    old = DEVICE
    DEVICE = dev
    try:
        yield
    finally:
        DEVICE = old

def probe_condition_outputs(st_items, deadline=None):
    """R / S / C probability vectors on a few items for the dtype gate. Stops early once `deadline` (unix time) has passed."""
    out = {"R": [], "S": [], "C": []}
    for it in st_items:
        if deadline is not None and out["R"] and time.time() > deadline:
            break
        ids = it["ids"] + RESP_IDS
        for k, lg in (("R", lm_logits(RECV, ids)), ("S", lm_logits(SHAR, ids)), ("C", c2c_logits(ids, ids, it["L"]))):
            out[k].append(to_pred(lg))
    return out

def mem_available_gb():
    """Host RAM available to this process in GiB: min(MemAvailable, cgroup limit - cgroup usage). None if unknown."""
    vals = []
    try:
        for line in open("/proc/meminfo"):
            if line.startswith("MemAvailable:"):
                vals.append(int(line.split()[1]) / 2**20)
    except Exception:
        pass
    for lim_p, cur_p in (("/sys/fs/cgroup/memory.max", "/sys/fs/cgroup/memory.current"),
                         ("/sys/fs/cgroup/memory/memory.limit_in_bytes", "/sys/fs/cgroup/memory/memory.usage_in_bytes")):
        try:
            lim = int(open(lim_p).read().strip())
            cur = int(open(cur_p).read().strip())
            if lim < 2**50:
                vals.append((lim - cur) / 2**30)
        except Exception:
            continue
    return min(vals) if vals else None

def fp32_fits_gpu():
    if not torch.cuda.is_available():
        return True
    total_gb = torch.cuda.get_device_properties(0).total_memory / 2**30
    return PAIR["fp32_gb"] < 0.80 * total_gb

def choose_dtype_and_load():
    want = os.environ.get("PCS_DTYPE") or CFG["DTYPE"]
    table_ = {"bf16": torch.bfloat16, "fp16": torch.float16, "fp32": torch.float32}
    if want != "auto":
        return build_stack(table_[want]), f"{want} (forced)"
    if SMOKE and not _flag("PCS_SMOKE_GATE"):
        return build_stack(torch.float32), "fp32 (smoke)"
    if NATIVE_BF16 and not SMOKE:
        return build_stack(torch.bfloat16), "bf16 (native on this GPU; same as the C2C repo)"
    # T4/P100: no fast bf16. Try fp16 and gate it against an fp32 reference on probe items.
    # Small pairs: the fp32 reference runs on the GPU. Large pairs (fp32 stack > 80% of GPU memory): it runs on the CPU, on fewer items.
    force_cpu_ref = _flag("PCS_SMOKE_CPU_REF")      # smoke-test hook: exercise the CPU-reference branch with tiny models
    on_gpu = (fp32_fits_gpu() or SMOKE) and not force_cpu_ref
    ref_dev = DEVICE if on_gpu else torch.device("cpu")
    probe_n = CFG["GATE_PROBES_GPU"] if on_gpu else CFG["GATE_PROBES_CPU"]
    cap_sec = float(os.environ.get("PCS_GATE_CAP_SEC", CFG["GATE_CPU_CAP_SEC"]))
    ref, pre_pool, ref_note = None, None, ""
    try:
        t_ref = time.time()
        if not on_gpu and not SMOKE:     # a CPU fp32 stack that does not fit RAM would be OOM-killed (no exception, run lost): check first
            avail = mem_available_gb()
            print(f"host RAM available: {avail if avail is None else round(avail, 1)} GiB; fp32 reference stack needs about {PAIR['fp32_gb']} GiB (+3 headroom)", flush=True)
            if avail is not None and avail < PAIR["fp32_gb"] + 3:
                raise MemoryError(f"only {avail:.1f} GiB host RAM available for a ~{PAIR['fp32_gb']} GiB fp32 reference stack")
        with on_device(ref_dev):
            st32 = build_stack(torch.float32, ref_dev); use_stack(st32)
            pre_pool = [finish_item(r) for r in seeded_sample(POOLS["obqa"], probe_n, "gate")[0]]
            ref = probe_condition_outputs(pre_pool, deadline=None if on_gpu else time.time() + cap_sec)
            free_stack(st32)
        n_ref = len(ref["R"])
        ref_note = f"fp32 reference on {ref_dev.type}: {n_ref}/{probe_n} probe items in {time.time() - t_ref:.0f}s"
        if n_ref < CFG["GATE_MIN_REF"] and (not SMOKE or force_cpu_ref):
            ref_note += " (too few: reference unusable)"
            ref = None
    except Exception as e:   # e.g. CPU out of memory while holding the fp32 stack
        ref_note = f"fp32 reference failed: {repr(e)[:200]}"
        ref = None
        try:
            free_stack(st32)
        except Exception:
            pass
    print(ref_note, flush=True)
    report, ok, st16 = {}, True, None
    try:
        st16 = build_stack(torch.float16); use_stack(st16)
        if pre_pool is None:
            pre_pool = [finish_item(r) for r in seeded_sample(POOLS["obqa"], probe_n, "gate")[0]]
        got = probe_condition_outputs(pre_pool if ref is None else pre_pool[:len(ref["R"])])
        for k in ("R", "S", "C"):
            fin = all(g["a"] != "?" for g in got[k])
            ok &= fin
            if ref is None:
                report[k] = dict(finite=fin, argmax_agree=None, mean_abs_dprob=None)
                continue
            agree = float(np.mean([g["a"] == r["a"] for g, r in zip(got[k], ref[k])]))
            dp = float(np.mean([np.abs(np.array(g["p"], float) - np.array(r["p"], float)).mean() for g, r in zip(got[k], ref[k]) if g["a"] != "?"])) if fin else float("nan")
            report[k] = dict(finite=fin, argmax_agree=round(agree, 3), mean_abs_dprob=round(dp, 4))
            ok &= agree >= 0.80 and dp <= 0.10
    except Exception as e:   # any fp16 failure (kernel, dtype mismatch, OOM) -> fall back
        report, ok = {"error": repr(e)[:300]}, False
    print("fp16 gate vs fp32 reference:", report, flush=True)
    if ok:
        if ref is None:
            DEVIATIONS.append(f"fp32 reference for the fp16 gate unavailable ({ref_note}); fp16 gated on finiteness of R/S/C logits only")
        return st16, "fp16 (passed sanity gate vs fp32)" if ref is not None else "fp16 (finite; no fp32 reference available)"
    if st16 is not None:
        free_stack(st16)
    if fp32_fits_gpu() or SMOKE:
        DEVIATIONS.append(f"fp16 failed the sanity gate {report}; fell back to fp32")
        return build_stack(torch.float32), "fp32 (fp16 failed sanity gate)"
    DEVIATIONS.append(f"fp16 failed the sanity gate {report}; fp32 does not fit this GPU, fell back to bf16 (emulated on pre-Ampere GPUs, slow)")
    return build_stack(torch.bfloat16), "bf16 (fp16 failed sanity gate; fp32 does not fit; emulated)"

t = time.time()
STACK, DTYPE_NOTE = choose_dtype_and_load()
use_stack(STACK)
COUNTERS["nonfinite"] = 0
print(f"models + fuser ready in {time.time()-t:.0f}s | dtype: {DTYPE_NOTE} | fuser projectors: {STACK['n_proj']}")
if torch.cuda.is_available():
    print(f"GPU memory after load: {torch.cuda.memory_allocated()/2**30:.1f} GiB allocated of {torch.cuda.get_device_properties(0).total_memory/2**30:.1f}")

POOL, RESERVE = build_items()
print(f"items: pool={len(POOL)} reserve={len(RESERVE)} | median prompt tokens={int(np.median([i['L'] for i in POOL]))} max={max(i['L'] for i in POOL)}")

# bank of same-length partner prompts for the mismatched-sharer control (everything loaded, tokenized lazily by length)
_bank_rows = POOLS["mmlu"] + POOLS["arc"] + POOLS["obqa"]
BANK = {}
for _r in _bank_rows:
    _ids = encode_user(TOK_R, make_prompt(_r["question"], _r["choices"]))
    BANK.setdefault(len(_ids), []).append((f"{_r['bench']}:{_r['uid']}", _ids))
print(f"partner bank: {sum(len(v) for v in BANK.values())} prompts over {len(BANK)} distinct lengths")

def pick_partners(it, K=1):
    """Up to K different wrong-question partners: prompts of exactly the same token length (receiver-template tokens), never the item
    itself, with distinct token sequences, drawn without replacement by a seeded shuffle per item (so the k-th wrong state differs from the others)."""
    cands = [(pid, ids) for pid, ids in BANK.get(it["L"], []) if pid != it["id"] and ids != it["ids"]]
    random.Random(f"{SEED}:mmK:{it['id']}").shuffle(cands)
    out, seen = [], set()
    for pid, ids in cands:
        key = tuple(ids)
        if key in seen:
            continue
        seen.add(key)
        out.append((pid, ids))
        if len(out) == K:
            break
    return out

def pick_partner(it):
    p = pick_partners(it, 1)
    return p[0] if p else None

def timeit(fn, n=6):
    fn()  # warm-up
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    t0 = time.time()
    for _ in range(n):
        fn()
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    return (time.time() - t0) / n

_it = sorted(POOL, key=lambda i: abs(i["L"] - 250))[0]
_ids = _it["ids"] + RESP_IDS
t_lm = timeit(lambda: lm_logits(RECV, _ids))
t_lm_s = timeit(lambda: lm_logits(SHAR, _ids))
t_c2c = timeit(lambda: c2c_logits(_ids, _ids, _it["L"]))
t_gen_r = timeit(lambda: hf_generate(RECV, TOK_R, _it["ids"], GEN_TOKENS, EOS_R, PAD_R), n=2)
t_gen_c = timeit(lambda: c2c_generate(_it["ids"], GEN_TOKENS), n=2)
N_B = len(POOL)
est_stage1 = N_B * (t_lm + 2 * t_lm_s + (1 + CFG["K_MM"]) * t_c2c)         # R, S, S_nat (skipped when identical), C and K C_mm on every item
est_stage2 = 0.5 * N_B * (2 * t_lm + (0.5 if CFG["RUN_T2_RATIONALE"] else 0))   # T and T2 on the disagreement set (about half of the items, upper bound)
est_C = CFG["N_PLANT"] * (2 * t_c2c + 2 * t_lm_s + 3 * t_lm)
est_A = CFG["N_PLUMB"] * (t_lm + t_lm_s + 2 * t_c2c + (2 * t_gen_r + 2 * t_gen_c if CFG["RUN_GEN_READOUT_IN_A"] else 0))
print(f"timing (median-length prompt): receiver fwd {t_lm*1e3:.0f} ms | sharer fwd {t_lm_s*1e3:.0f} ms | C2C fwd {t_c2c*1e3:.0f} ms | HF gen64 {t_gen_r:.2f}s | C2C gen64 {t_gen_c:.2f}s")
print(f"ESTIMATE  Part A ~{est_A/60:.0f} min | Part B ~{(est_stage1+est_stage2)/60:.0f} min | Part C ~{est_C/60:.0f} min  (heuristic; includes no model-download time)")
if (est_A + est_stage1 + est_stage2) / 60 > CFG["TIME_BUDGET_MIN"] and not (SMOKE or QUICK):
    scale = (CFG["TIME_BUDGET_MIN"] * 60 - est_A) / (est_stage1 + est_stage2)
    keep = max(600, int(len(POOL) * scale))
    DEVIATIONS.append(f"estimated A+B runtime exceeded {CFG['TIME_BUDGET_MIN']} min; Part B pool cut from {len(POOL)} to {keep} items (pre-registered order)")
    POOL = POOL[:keep]
    print("!! time budget: Part B pool reduced to", keep)
```

```python
# --- 11. Part A: plumbing (OpenBookQA; generation readout = the paper's protocol, and the logit readout) + mismatched-sharer under generation ---
RES = {"config": CFG, "deviations": DEVIATIONS, "pair": {"key": PAIR_KEY, "receiver": RECV_ID, "sharer": SHAR_ID, "fuser": f"{FUSER_REPO}/{FUSER_SUB}/final"},
       "partA": None, "partB": None, "partB_run3": None, "partC": None, "numeric": None}
PLUMBING_OK = None

def gen_mismatch_stats(c_gen, m_gen, r_gen, y, has_mm):
    """PREREG_run2 (c), kept in run 3: C_mm_gen vs C_gen on the Part A items that have a same-length partner (generation readout; letters or None)."""
    y = np.asarray(y)
    hm = np.asarray(has_mm, bool)
    cC = np.array([a == b for a, b in zip(c_gen, y)], float)
    cM = np.array([a == b for a, b in zip(m_gen, y)], float)
    cR = np.array([a == b for a, b in zip(r_gen, y)], float)
    agree = np.array([(a is not None) and a == b for a, b in zip(c_gen, m_gen)], float)
    return {"n_with_partner": int(hm.sum()), "coverage": float(hm.mean()),
            "acc": {"C": fmt(boot_ci(cC[hm]), True), "C_mm": fmt(boot_ci(cM[hm]), True)},
            "acc_point": {"C": float(cC[hm].mean()) if hm.any() else None, "C_mm": float(cM[hm].mean()) if hm.any() else None},
            "C_minus_C_mm": fmt(boot_ci((cC - cM)[hm])), "C_minus_C_mm_ci": boot_ci((cC - cM)[hm]),
            "C_mm_minus_R": fmt(boot_ci((cM - cR)[hm])),
            "agreement_C_vs_C_mm": boot_ci(agree[hm]),
            "unparsed_C_mm": int(sum(a is None for a, h in zip(m_gen, hm) if h))}

def run_part_A():
    global PLUMBING_OK
    obqa, _ = seeded_sample(POOLS["obqa"], CFG["N_PLUMB"], "partA")
    items = [finish_item(r) for r in obqa]
    y = np.array([it["y"] for it in items])
    rows = {"gen": {"R": [], "S": [], "C": [], "M": []}, "logit": {"R": [], "S": [], "C": []}}
    has_mm = []
    chk = {"norproj_vs_recv_agree": [], "norproj_dprob": [], "list_vs_tensor_maxdiff": []}
    t_a = time.time()
    for n_, it in enumerate(items):
        ids_full, ids_nat_full = it["ids"] + RESP_IDS, it["ids_nat"] + RESP_IDS
        r_lg = lm_logits(RECV, ids_full)
        r_pred = to_pred(r_lg)
        rows["logit"]["R"].append(r_pred["a"])
        rows["logit"]["S"].append(to_pred(lm_logits(SHAR, ids_nat_full))["a"])    # sharer-alone with its own template, as in C2C's baseline
        c_lg = c2c_logits(ids_full, ids_full, it["L"])
        rows["logit"]["C"].append(to_pred(c_lg)["a"])
        off_lg = c2c_logits(ids_full, ids_full, it["L"], sharer_mask=-1)           # projection OFF must reproduce the plain receiver
        off_pred = to_pred(off_lg)
        chk["norproj_vs_recv_agree"].append(float(off_pred["a"] == r_pred["a"]))
        if off_pred["a"] != "?" and r_pred["a"] != "?":
            chk["norproj_dprob"].append(float(np.abs(np.array(off_pred["p"]) - np.array(r_pred["p"])).mean()))
        if n_ < 10:
            chk["list_vs_tensor_maxdiff"].append(float((c2c_logits(ids_full, ids_full, it["L"], as_tensor=True).float() - c_lg.float()).abs().max()))
        if CFG["RUN_GEN_READOUT_IN_A"]:
            rows["gen"]["R"].append(extract_answer_from_content(hf_generate(RECV, TOK_R, it["ids"], GEN_TOKENS, EOS_R, PAD_R)))
            rows["gen"]["S"].append(extract_answer_from_content(hf_generate(SHAR, TOK_S, it["ids_nat"], GEN_TOKENS, EOS_S, PAD_S)))
            rows["gen"]["C"].append(extract_answer_from_content(c2c_generate(it["ids"], GEN_TOKENS)))
            part = pick_partner(it)      # PREREG_run2 (c), kept in run 3: mismatched sharer (cache of a different same-length prompt) under the generation readout
            has_mm.append(part is not None)
            rows["gen"]["M"].append(extract_answer_from_content(c2c_generate(it["ids"], GEN_TOKENS, ids_s=part[1])) if part is not None else None)
        if (n_ + 1) % 50 == 0:
            print(f"  part A {n_+1}/{len(items)}  {time.time()-t_a:.0f}s", flush=True)
    RES["partA_raw"] = {"ids": [it["id"] for it in items], "y": y.tolist(), "rows": rows, "has_mm": has_mm, "checks": chk}   # kept even if the analysis below fails
    out = {"n": len(items), "paper_obqa_acc": PAPER_OBQA, "readouts": {}, "checks": {}}
    corr_all = {}
    for ro in ("gen", "logit"):
        if not rows[ro]["R"]:
            continue
        corr = {k: np.array([a == b for a, b in zip(rows[ro][k], y)], float) for k in "RSC"}
        corr_all[ro] = corr
        out["readouts"][ro] = {
            "acc": {k: fmt(boot_ci(corr[k]), pct=True) for k in "RSC"},
            "acc_point": {k: float(corr[k].mean()) for k in "RSC"},
            "acc_ci": {k: boot_ci(corr[k]) for k in "RSC"},
            "C_minus_R": fmt(boot_ci(corr["C"] - corr["R"])),
            "C_minus_S": fmt(boot_ci(corr["C"] - corr["S"])),
            "C_minus_R_ci": boot_ci(corr["C"] - corr["R"]),
            "unparsed": {k: int(sum(a is None for a in rows[ro][k])) for k in "RSC"},
        }
    if rows["gen"]["M"]:   # mismatched sharer under the generation readout, on the items that have a same-length partner
        out["gen_mismatch"] = gen_mismatch_stats(rows["gen"]["C"], rows["gen"]["M"], rows["gen"]["R"], y, has_mm)
    out["checks"] = {"projection_off_matches_receiver": float(np.mean(chk["norproj_vs_recv_agree"])),
                     "projection_off_mean_abs_dprob": float(np.mean(chk["norproj_dprob"])) if chk["norproj_dprob"] else None,
                     "list_vs_tensor_max_logit_diff": float(max(chk["list_vs_tensor_maxdiff"])) if chk["list_vs_tensor_maxdiff"] else None,
                     "nonfinite_so_far": COUNTERS["nonfinite"]}
    out["per_item"] = [{"id": it["id"], "y": it["y"], **{f"{ro}_{k}": rows[ro][k][i] for ro in rows for k in rows[ro] if rows[ro][k]}} for i, it in enumerate(items)]
    ref_ro = "gen" if "gen" in out["readouts"] else "logit"
    lo = out["readouts"][ref_ro]["C_minus_R_ci"][1]
    parse_ok = True
    if "gen" in out["readouts"]:
        parse_ok = all(v <= 0.10 * len(items) for v in out["readouts"]["gen"]["unparsed"].values()) and \
                   out.get("gen_mismatch", {}).get("unparsed_C_mm", 0) <= 0.10 * len(items)
    # PREREG_run3 plumbing = PREREG.md's three criteria (the paper's Table 9 reports this pair, so the original "C2C beats the receiver" criterion is restored)
    # plus the parse-rate check of run 2. The paper-numbers comparison is informational (CI at N=200 is about +-7 pts).
    out["criteria"] = {"no_nonfinite_logits": COUNTERS["nonfinite"] == 0,
                       f"C2C_beats_receiver_CI_excludes_0[{ref_ro}]": bool(lo > 0),
                       "projection_off_matches_receiver>=95%": bool(out["checks"]["projection_off_matches_receiver"] >= 0.95),
                       "generation_parse_rate>=90%_for_R,S,C,C_mm": bool(parse_ok)}
    out["informational"] = {}
    if PAPER_OBQA and "gen" in out["readouts"]:
        cis = out["readouts"]["gen"]["acc_ci"]
        out["informational"]["paper_value_inside_our_95CI[gen]"] = {k: bool(cis[k][1] * 100 <= PAPER_OBQA[k] <= cis[k][2] * 100) for k in "RSC"}
    PLUMBING_OK = all(out["criteria"].values())
    out["plumbing_ok"] = PLUMBING_OK
    out["seconds"] = round(time.time() - t_a)
    return out

def print_part_A(a):
    if PAPER_OBQA:
        print("\n=== PART A: plumbing (OpenBookQA, N=%d; paper N=500: R %.1f / S %.1f / C2C %.1f) ===" % (a["n"], *[PAPER_OBQA[k] for k in "RSC"]))
    else:
        print("\n=== PART A: plumbing (OpenBookQA, N=%d; the paper reports no number for this pair) ===" % a["n"])
    rows = []
    for ro, d in a["readouts"].items():
        for k, nm in (("R", "receiver alone"), ("S", "sharer alone"), ("C", "C2C fused")):
            rows.append([ro, nm, d["acc"][k], f"{PAPER_OBQA[k]:.1f}%" if PAPER_OBQA else "-", d["unparsed"][k]])
    table(rows, ["readout", "condition", "accuracy [95% CI]", "paper", "unparsed"])
    for ro, d in a["readouts"].items():
        print(f"  [{ro}] C2C - receiver: {d['C_minus_R']}   C2C - sharer: {d['C_minus_S']}")
    g = a.get("gen_mismatch")
    if g:
        print(f"  [gen] mismatched sharer (n={g['n_with_partner']} with a same-length partner): acc C {g['acc']['C']} | C_mm {g['acc']['C_mm']} | C - C_mm {g['C_minus_C_mm']} | answer agreement C vs C_mm {fmt(g['agreement_C_vs_C_mm'], True)} | unparsed C_mm {g['unparsed_C_mm']}")
    print("  checks:", a["checks"])
    print("  criteria (PREREG_run3):", a["criteria"], "| informational:", a["informational"])
    print("  PLUMBING:", "OK" if a["plumbing_ok"] else "!! NOT OK: see failed criterion above; Part B is not interpretable until it is understood", f"({a['seconds']}s)")

if CFG["RUN_A"]:
    try:
        RES["partA"] = run_part_A()
        print_part_A(RES["partA"])
        if not RES["partA"]["plumbing_ok"]:
            DEVIATIONS.append("Part A plumbing check did NOT pass; treat Parts B/C as uninterpretable")
    except Exception as e:   # never lose Part B because of a reporting bug in Part A; raw rows are in RES["partA_raw"]
        import traceback; traceback.print_exc()
        DEVIATIONS.append(f"Part A analysis raised {repr(e)[:200]}; raw rows saved in partA_raw; plumbing status unknown")
        PLUMBING_OK = None
```

```python
# --- 12. Part B: fidelity test (stage 1 on all items incl. the mismatched-sharer control C_mm; stage 2 text handoffs on disagreement items) ---
RECORDS = {}   # item id -> {cond: {"a": letter, "p": [4 probs], "lp": [4 log-probs]} | text fields}
ITEMS = {}
NUM_ROWS = []  # numeric task rows (filled in the numeric cell; checkpointed with the MC records)

def save_partial():
    try:
        PARTIAL_PATH.write_text(json.dumps({"records": RECORDS, "numeric_rows": NUM_ROWS, "config": CFG}))
    except Exception as e:
        print("could not write partial results:", e)

def stage1(items):
    t_s = time.time()
    for n_, it in enumerate(items):
        ITEMS[it["id"]] = it
        rec = RECORDS.setdefault(it["id"], {})
        if "C" in rec and "C_mm0" in rec:
            continue
        ids_full, ids_nat_full = it["ids"] + RESP_IDS, it["ids_nat"] + RESP_IDS
        rec["R"] = to_pred(lm_logits(RECV, ids_full))
        rec["S"] = to_pred(lm_logits(SHAR, ids_full))          # sharer fed the exact tokens the fuser feeds it
        if ids_nat_full == ids_full:                            # same tokenizer AND template (Qwen3 -> Qwen3): S_nat is identical, do not recompute
            rec["S_nat"] = rec["S"]
        else:
            rec["S_nat"] = to_pred(lm_logits(SHAR, ids_nat_full))   # sharer with its own chat template (sensitivity)
        rec["C"] = to_pred(c2c_logits(ids_full, ids_full, it["L"]))
        parts = pick_partners(it, CFG["K_MM"])                  # PREREG_run3 safeguard 1: K different wrong-question caches on ALL items
        for k in range(CFG["K_MM"]):
            if k < len(parts):
                rec[f"C_mm{k}"] = to_pred(c2c_logits(ids_full, parts[k][1] + RESP_IDS, it["L"]))
                rec[f"C_mm{k}"]["partner"] = parts[k][0]
            else:
                rec[f"C_mm{k}"] = None
        rec["C_mm"] = rec["C_mm0"]                              # the belief-transfer table (analyze_B) uses the first wrong state
        if (n_ + 1) % 100 == 0:
            print(f"  stage1 {n_+1}/{len(items)}  {time.time()-t_s:.0f}s  nonfinite={COUNTERS['nonfinite']}", flush=True)
            save_partial()

def stage2(items):
    """Only items where sharer and receiver disagree: text handoffs (the mismatched-sharer controls are computed in stage 1)."""
    d_items = [it for it in items if RECORDS[it["id"]]["S"]["a"] != RECORDS[it["id"]]["R"]["a"]
               and "?" not in (RECORDS[it["id"]]["S"]["a"], RECORDS[it["id"]]["R"]["a"])]
    t_s = time.time()
    for n_, it in enumerate(d_items):
        rec = RECORDS[it["id"]]
        s_letter = rec["S"]["a"]
        if "T" not in rec:
            rec["T"] = to_pred(lm_logits(RECV, encode_user(TOK_R, f"Another model answered: {s_letter}\n\n" + it["prompt"]) + RESP_IDS))
        if (n_ + 1) % 100 == 0:
            print(f"  stage2 {n_+1}/{len(d_items)}  {time.time()-t_s:.0f}s", flush=True)
            save_partial()
    if CFG["RUN_T2_RATIONALE"]:
        todo = [it for it in d_items if "T2" not in RECORDS[it["id"]]]
        if todo:
            prompts = []
            for it in todo:
                s_letter = RECORDS[it["id"]]["S"]["a"]
                text = TOK_S.apply_chat_template([{"role": "user", "content": it["prompt"]}], tokenize=False, add_generation_prompt=True)
                prompts.append(TOK_S(text + f"The correct answer is {s_letter}. Brief reason:", add_special_tokens=False)["input_ids"])
            gens = hf_generate_batch(SHAR, TOK_S, prompts, 40, EOS_S, PAD_S)
            for it, g in zip(todo, gens):
                why = g.strip().split("\n")[0].strip()
                m = re.search(r"^(.{15,}?[.!?])(\s|$)", why)
                why = (m.group(1) if m else why)[:240] or "(no reason given)"
                s_letter = RECORDS[it["id"]]["S"]["a"]
                RECORDS[it["id"]]["T2_text"] = why
                RECORDS[it["id"]]["T2"] = to_pred(lm_logits(RECV, encode_user(TOK_R, f"Another model answered: {s_letter}. Its reasoning: {why}\n\n" + it["prompt"]) + RESP_IDS))
    save_partial()

def arrays(ids, cond, key="a"):
    return np.array([RECORDS[i][cond][key] if RECORDS[i].get(cond) else "" for i in ids])

def analyze_B(ids):
    ids = [i for i in ids if all(k in RECORDS[i] for k in ("R", "S", "S_nat", "C"))]
    y = np.array([ITEMS[i]["y"] for i in ids])
    R, S, Sn, C = (arrays(ids, k) for k in ("R", "S", "S_nat", "C"))
    ok = (R != "?") & (S != "?") & (C != "?")
    D = ok & (S != R)
    cells = {"A": D & (S == y), "B": D & (S != y) & (R == y), "W": D & (S != y) & (R != y)}
    cells["B+W"] = cells["B"] | cells["W"]
    has = lambda cond: np.array([bool(RECORDS[i].get(cond)) and RECORDS[i][cond]["a"] != "?" for i in ids])
    C_mm, T, T2 = arrays(ids, "C_mm"), arrays(ids, "T"), arrays(ids, "T2")
    hm, ht, ht2 = has("C_mm"), has("T"), has("T2")
    out = {"n_items": len(ids), "n_disagree": int(D.sum()), "cell_sizes": {k: int(v.sum()) for k, v in cells.items()},
           "mismatch_partner_coverage_on_D": float(hm[D].mean()) if D.any() else None}
    out["accuracy_all"] = {k: fmt(boot_ci((a == y)[ok]), pct=True) for k, a in (("R", R), ("S", S), ("S_nat", Sn), ("C", C))}
    out["accuracy_all_point"] = {k: float((a == y)[ok].mean()) for k, a in (("R", R), ("S", S), ("S_nat", Sn), ("C", C))}
    out["accuracy_all"]["C_minus_R"] = fmt(boot_ci(((C == y).astype(float) - (R == y).astype(float))[ok]))
    out["accuracy_on_D"] = {k: fmt(boot_ci((a == y)[D & h]), pct=True) for k, a, h in
                            (("R", R, np.ones_like(D)), ("S", S, np.ones_like(D)), ("C", C, np.ones_like(D)),
                             ("C_mm", C_mm, hm), ("T", T, ht), ("T2", T2, ht2))}
    out["accuracy_on_D"]["C_minus_C_mm"] = fmt(boot_ci(((C == y).astype(float) - (C_mm == y).astype(float))[D & hm]))
    out["accuracy_on_D"]["C_minus_C_mm_ci"] = boot_ci(((C == y).astype(float) - (C_mm == y).astype(float))[D & hm])
    inh = {}
    for cn, cm in cells.items():
        row = {"n": int(cm.sum())}
        row["Inh_C"] = boot_ci((C == S)[cm]); row["Rev_C"] = boot_ci((C == R)[cm])
        row["Inh_C_other"] = boot_ci(((C != S) & (C != R))[cm])
        mm = cm & hm
        row["n_with_mm"] = int(mm.sum())
        row["Inh_C_mm"] = boot_ci((C_mm == S)[mm]); row["Inh_C_same_items"] = boot_ci((C == S)[mm])
        row["Rev_C_mm"] = boot_ci((C_mm == R)[mm]); row["Rev_C_same_items"] = boot_ci((C == R)[mm])
        row["E_excess"] = boot_ci(((C == S).astype(float) - (C_mm == S).astype(float))[mm])
        row["Inh_T"] = boot_ci((T == S)[cm & ht]); row["Inh_T2"] = boot_ci((T2 == S)[cm & ht2])
        row["Inh_C_native_S"] = boot_ci((C == Sn)[cm & (Sn != "?")])
        inh[cn] = row
    out["inheritance"] = inh
    out["asymmetry_Inh_C_A_minus_B"] = boot_ci_unpaired_diff((C == S)[cells["A"]], (C == S)[cells["B"]])
    eB = inh["B"]["E_excess"]; tB = inh["B"]["Inh_T"]
    out["bridge_over_text_B"] = (eB[0] / tB[0]) if tB[0] and tB[0] > 0 else None
    out["S_native_agreement"] = float((S == Sn)[ok].mean())
    out["letter_marginals"] = {k: {l: float((a == l).mean()) for l in LETTERS} for k, a in (("R", R), ("S", S), ("C", C))}
    mm_leak = [(RECORDS[i]["C_mm"]["a"], RECORDS[RECORDS[i]["C_mm"]["partner"]]["S"]["a"]) for i in ids
               if RECORDS[i].get("C_mm") and RECORDS[i]["C_mm"].get("partner") in RECORDS and "S" in RECORDS[RECORDS[i]["C_mm"]["partner"]]]
    out["mismatch_letter_leak_vs_partner_sharer_answer"] = {"n": len(mm_leak), "rate": float(np.mean([a == b for a, b in mm_leak])) if mm_leak else None, "chance": 0.25}
    out["reading"] = auto_reading(out)
    return out

def auto_reading(o):
    B_, A_ = o["inheritance"]["B"], o["inheritance"]["A"]
    eB, eA = B_["E_excess"], A_["E_excess"]
    if o["cell_sizes"]["A"] < CFG["MIN_CELL"] or o["cell_sizes"]["B"] < CFG["MIN_CELL"]:
        return f"UNDERPOWERED: |A|={o['cell_sizes']['A']} |B|={o['cell_sizes']['B']} (< {CFG['MIN_CELL']}); no claim (PREREG)"
    if eB[1] >= 0.10:
        return "BELIEF TRANSFER: receiver abandons a correct answer for the sharer's wrong one, >=10pp beyond the mismatched-sharer null (lower CI of E_B >= 0.10)"
    if eB[2] < 0.10 and eA[0] >= 0.10:
        return "TRUTH-BIASED HELPER: follows the sharer when it is right (E_A >= 0.10) but not when it is wrong (upper CI of E_B < 0.10)"
    d = o["accuracy_on_D"]["C_minus_C_mm_ci"]
    if eA[1] >= -0.05 and eA[2] <= 0.05 and eB[1] >= -0.05 and eB[2] <= 0.05 and d[1] <= 0 <= d[2]:
        return "NON-SPECIFIC PERTURBATION: inheritance is indistinguishable from the mismatched-sharer null and accuracy does not depend on the sharer's content"
    return "MIXED / UNDERPOWERED: matches no pre-registered row; no claim"

# ---- PREREG_run2 analysis: calibrated-receiver baseline (a), mismatched control on all items (b), thresholds (e) ----
def _lp_vec(pred):
    """Log-softmax over the four option letters (stored as `lp`; recomputed from `p` if absent)."""
    if pred.get("lp") and pred["lp"][0] is not None:
        return np.array(pred["lp"], float)
    p = np.array(pred["p"], float)
    return np.log(np.clip(p, 1e-12, None))

def cal_folds(ids, seed=None):
    """Deterministic 2-fold split of the scored items; the SAME split is used for every condition (R, C, C_mm)."""
    seed = CFG["CAL_SEED"] if seed is None else seed
    order = sorted(ids)
    perm = list(range(len(order)))
    random.Random(f"{seed}:calfold").shuffle(perm)
    fold = {}
    for rank, j in enumerate(perm):
        fold[order[j]] = rank % 2
    return fold

def crossfit_calibrate(ids, cond, fold=None):
    """PREREG_run2 (a), kept in run 3. Label-free per-letter prior correction, 2-fold cross-fit: items in fold k are scored as
    argmax_l [ logp_l - mean_over_items_of_the_OTHER_fold(logp_l) ] (ties -> lower letter). Uses no labels and no sharer
    information beyond what `cond` itself outputs. Returns {id: letter} for items where `cond` is present and finite."""
    fold = fold if fold is not None else cal_folds(ids)
    have = [i for i in ids if RECORDS[i].get(cond) and RECORDS[i][cond]["a"] != "?"]
    lp = {i: _lp_vec(RECORDS[i][cond]) for i in have}
    out = {}
    for k in (0, 1):
        train = [lp[i] for i in have if fold[i] != k]
        if not train:
            continue
        prior = np.mean(train, axis=0)
        for i in have:
            if fold[i] == k:
                out[i] = LETTERS[int(np.argmax(lp[i] - prior))]
    return out

def cal_array(ids, cond, fold):
    d = crossfit_calibrate(ids, cond, fold)
    return np.array([d.get(i, "?") for i in ids])

def analyze_run3(ids, K=None):
    """MC analysis of PREREG_run3: K wrong-question controls on every item, calibrated receiver, thresholds. Primary quantity
    Delta_specific = acc(C) - mean_k acc(C_mm,k) (paired by item) on the items that have all K partners."""
    K = CFG["K_MM"] if K is None else K
    ids = [i for i in ids if all(k in RECORDS[i] for k in ("R", "S", "C"))]
    y = np.array([ITEMS[i]["y"] for i in ids])
    fold = cal_folds(ids)
    R, S, C = (arrays(ids, k) for k in ("R", "S", "C"))
    Cmk = [arrays(ids, f"C_mm{k}") for k in range(K)]
    Rcal, Ccal = cal_array(ids, "R", fold), cal_array(ids, "C", fold)
    Cmcalk = [cal_array(ids, f"C_mm{k}", fold) for k in range(K)]
    ok = (R != "?") & (S != "?") & (C != "?")
    hk = [~np.isin(a, ["", "?"]) for a in Cmk]
    hall = np.logical_and.reduce(hk)                 # item has all K same-length wrong-question partners with valid predictions
    okK = ok & hall
    D = ok & (S != R)
    DK = D & hall
    cells = {"A": D & (S == y), "B": D & (S != y) & (R == y), "W": D & (S != y) & (R != y)}
    corr = lambda a: (a == y).astype(float)
    acc = lambda a, m: boot_ci(corr(a)[m])
    dacc = lambda a, b, m: boot_ci((corr(a) - corr(b))[m])
    mean_cm = np.mean([corr(a) for a in Cmk], axis=0)
    mean_cmcal = np.mean([corr(a) for a in Cmcalk], axis=0)
    out = {"K": K, "n_items": int(ok.sum()), "n_disagree": int(D.sum()), "n_K_complete": int(okK.sum()), "n_K_complete_D": int(DK.sum()),
           "coverage_K_all": float(hall[ok].mean()) if ok.any() else None, "coverage_K_D": float(hall[D].mean()) if D.any() else None,
           "coverage_any_partner": float(np.logical_or.reduce(hk)[ok].mean()) if ok.any() else None,
           "calibration": {"method": "label-free per-letter mean log-prob prior, 2-fold cross-fit", "seed": CFG["CAL_SEED"], "n_per_fold": [int(sum(1 for v in fold.values() if v == k)) for k in (0, 1)]}}
    pts = {"R": R, "Rcal": Rcal, "S": S, "C": C, "Ccal": Ccal}
    out["accuracy_all"] = {k: fmt(acc(a, ok), True) for k, a in pts.items()}
    out["accuracy_all_point"] = {k: float(corr(a)[ok].mean()) for k, a in pts.items()}
    sub = {"R": R, "Rcal": Rcal, "C": C, "Ccal": Ccal}
    sub.update({f"C_mm{k}": a for k, a in enumerate(Cmk)})
    sub.update({f"C_mm{k}_cal": a for k, a in enumerate(Cmcalk)})
    out["accuracy_K_complete"] = {k: fmt(acc(a, okK), True) for k, a in sub.items()}
    out["accuracy_K_complete_point"] = {k: float(corr(a)[okK].mean()) if okK.any() else None for k, a in sub.items()}
    out["accuracy_K_complete_point"]["mean_C_mm"] = float(mean_cm[okK].mean()) if okK.any() else None
    out["accuracy_K_complete_point"]["mean_C_mm_cal"] = float(mean_cmcal[okK].mean()) if okK.any() else None
    dd = dict(pts); dd.update({"C_mm_mean": None})
    out["accuracy_on_D"] = {k: fmt(acc(a, D), True) for k, a in pts.items()}
    out["accuracy_on_D"]["C_mm_mean"] = fmt(boot_ci(mean_cm[DK]), True)
    cmp_ = {
        "Delta_specific": boot_ci((corr(C) - mean_cm)[okK]),            # PRIMARY: acc(C) - mean_k acc(C_mm,k), all K-complete items
        "Delta_specific_D": boot_ci((corr(C) - mean_cm)[DK]),
        "Delta_specific_cal": boot_ci((corr(Ccal) - mean_cmcal)[okK]),   # both calibrated (each C_mm,k by its own cross-fit prior)
        "Delta_specific_cal_D": boot_ci((corr(Ccal) - mean_cmcal)[DK]),
        "C_minus_R_all": dacc(C, R, ok),
        "Rcal_minus_R_all": dacc(Rcal, R, ok),
        "C_minus_Rcal_all": dacc(C, Rcal, ok),                          # PRIMARY (a)
        "C_minus_Rcal_D": dacc(C, Rcal, D),
        "Ccal_minus_Rcal_all": dacc(Ccal, Rcal, ok),
        "Ccal_minus_Rcal_D": dacc(Ccal, Rcal, D),
        "Cmm_mean_minus_R_all": boot_ci((mean_cm - corr(R))[okK]),
        "Cmm_mean_minus_Rcal_all": boot_ci((mean_cm - corr(Rcal))[okK]),
    }
    for k in range(K):
        cmp_[f"C_minus_Cmm{k}_all"] = dacc(C, Cmk[k], okK)
    out["comparisons"] = cmp_
    pC = np.array([RECORDS[i]["C"]["p"] if RECORDS[i].get("C") and RECORDS[i]["C"]["a"] != "?" else [np.nan] * 4 for i in ids], float)
    per_k, dps = [], []
    for k in range(K):
        pM = np.array([RECORDS[i][f"C_mm{k}"]["p"] if RECORDS[i].get(f"C_mm{k}") and RECORDS[i][f"C_mm{k}"]["a"] != "?" else [np.nan] * 4 for i in ids], float)
        dp = np.abs(pC - pM).mean(1)
        per_k.append({"k": k, "acc_C_mm": fmt(acc(Cmk[k], okK), True), "C_minus_Cmm": cmp_[f"C_minus_Cmm{k}_all"],
                      "agreement_all": {"n": int(okK.sum()), "same_answer": boot_ci((C == Cmk[k])[okK]), "mean_abs_dprob": float(np.nanmean(dp[okK])) if okK.any() else None},
                      "agreement_D": {"n": int(DK.sum()), "same_answer": boot_ci((C == Cmk[k])[DK])},
                      "agreement_cal_all": boot_ci((Ccal == Cmcalk[k])[okK]), "agreement_cal_D": boot_ci((Ccal == Cmcalk[k])[DK]),
                      "agreement_by_cell": {c_: boot_ci((C == Cmk[k])[m & hall]) for c_, m in cells.items()}})
    out["per_k"] = per_k
    pts_k = [p_["C_minus_Cmm"][0] for p_ in per_k]
    accs_k = [float(corr(a)[okK].mean()) if okK.any() else float("nan") for a in Cmk]
    pair_ag = [float((Cmk[a] == Cmk[b])[okK].mean()) for a in range(K) for b in range(a + 1, K)] if okK.any() else []
    out["spread_across_k"] = {"acc_C_mm_points": accs_k, "acc_C_mm_range": float(max(accs_k) - min(accs_k)) if accs_k else None,
                              "acc_C_mm_sd": float(np.std(accs_k)) if accs_k else None,
                              "C_minus_Cmm_points": pts_k, "C_minus_Cmm_range": float(max(pts_k) - min(pts_k)) if pts_k else None,
                              "wrong_states_pairwise_same_answer": pair_ag}
    out["letter_marginals"] = {k: {l: float((a[ok] == l).mean()) for l in LETTERS} for k, a in (("R", R), ("Rcal", Rcal), ("S", S), ("C", C), ("Ccal", Ccal), ("C_mm0", Cmk[0]))}
    out["reading_mc_only"], out["reading_flags_mc"] = auto_reading_run3(out, None)
    return out

def auto_reading_run3(mc, num=None):
    """PREREG_run3 decision rule. MC: Delta_specific = acc(C) - mean_k acc(C_mm,k); calibrated receiver; numeric task (verdict from
    acc(C) - acc(C_mm) on GSM8K: positive = lower CI > 0, null = CI includes 0, negative = upper CI < 0, unavailable = N < 60 or not run).
    CONTENT-SPECIFIC: lower CI Delta >= +0.02 AND C beats calibrated R (lower CI > 0) AND numeric positive.
    PARTIAL: not content-specific, but lower CI of Delta > 0 (below +0.02, or MC specific without numeric corroboration) OR numeric positive
             while MC is not specific (the tasks disagree): 'some genuine transfer; accuracy overstates it', with magnitudes.
    NON-SPECIFIC: CI of Delta includes 0 AND calibrated R within 2 pts of C (point estimates) (numeric not positive).
    Otherwise MIXED / NO CLAIM."""
    c = mc["comparisons"]
    d, dc = c["Delta_specific"], c["C_minus_Rcal_all"]
    numv = (num or {}).get("verdict", "unavailable")
    if d[0] != d[0] or dc[0] != dc[0]:
        return "NO CLAIM: no complete set of wrong-question controls or no calibrated-receiver data", {}
    flags = {"Delta_lowerCI>=+0.02": bool(d[1] >= 0.02), "Delta_lowerCI>0": bool(d[1] > 0), "C_beats_calibratedR_lowerCI>0": bool(dc[1] > 0),
             "Delta_CI_includes_0": bool(d[1] <= 0 <= d[2]), "calibratedR_within_2pts_of_C": bool(abs(dc[0]) <= 0.02), "numeric_verdict": numv}
    mag = f"Delta_specific {d[0]:+.3f} [{d[1]:+.3f},{d[2]:+.3f}]"
    if num and num.get("C_minus_Cmm_ci"):
        nd = num["C_minus_Cmm_ci"]
        mag += f"; numeric C - C_mm {nd[0]:+.3f} [{nd[1]:+.3f},{nd[2]:+.3f}] (n={num['n']})"
    if flags["Delta_lowerCI>=+0.02"] and flags["C_beats_calibratedR_lowerCI>0"] and numv == "positive":
        return f"CONTENT-SPECIFIC COMMUNICATION: MC and numeric tasks both show it, and C2C beats the calibrated receiver. {mag}", flags
    if flags["Delta_lowerCI>0"] or numv == "positive":
        return f"PARTIAL: some genuine transfer; accuracy overstates it. {mag}", flags
    if flags["Delta_CI_includes_0"] and flags["calibratedR_within_2pts_of_C"]:
        return f"NON-SPECIFIC: the sharer's cache for this question adds nothing detectable and a calibrated receiver is within 2 pts of C2C. {mag}", flags
    return f"MIXED / NO CLAIM: matches no pre-registered row of PREREG_run3. {mag}", flags

def print_run3(o):
    print(f"\n=== PART B, PREREG_run3: K={o['K']} wrong-question controls on ALL items, calibrated receiver (logit readout) ===")
    print(f"items scored: {o['n_items']} | with all K partners: {o['n_K_complete']} ({100*(o['coverage_K_all'] or 0):.1f}%; any partner {100*(o['coverage_any_partner'] or 0):.1f}%) | D (S != R): {o['n_disagree']} | calibration: {o['calibration']['method']}, seed {o['calibration']['seed']}")
    table([[k, v] for k, v in o["accuracy_all"].items()], ["accuracy, all scored items", "[95% CI]"])
    print()
    table([[k, v] for k, v in o["accuracy_K_complete"].items()], ["accuracy, K-complete items", "[95% CI]"])
    print()
    table([[k, v] for k, v in o["accuracy_on_D"].items()], ["accuracy on D", "[95% CI]"])
    print("\npaired differences (bootstrap 95% CI):")
    table([[k, fmt(v)] for k, v in o["comparisons"].items()], ["comparison", "difference [95% CI]"])
    print("\nper wrong state k: answer agreement with C (same letter)")
    table([[p_["k"], p_["acc_C_mm"], fmt(p_["C_minus_Cmm"]), fmt(p_["agreement_all"]["same_answer"], True), fmt(p_["agreement_D"]["same_answer"], True),
            fmt(p_["agreement_cal_all"], True), round(p_["agreement_all"]["mean_abs_dprob"] or 0, 4)] for p_ in o["per_k"]],
          ["k", "acc C_mm,k", "C - C_mm,k", "agree (all)", "agree (D)", "agree after calibration", "mean |dprob|"])
    print("spread across k:", {k: (round(v, 4) if isinstance(v, float) else [round(x, 4) for x in v]) for k, v in o["spread_across_k"].items()})
    print("predicted-letter marginals:", {k: {l: round(x, 2) for l, x in v.items()} for k, v in o["letter_marginals"].items()})
    print("\nMC-ONLY READING (PREREG_run3):", o["reading_mc_only"], o["reading_flags_mc"])

def print_part_B(b):
    print("\n=== PART B: fidelity (logit readout; D = items where sharer-alone != receiver-alone) ===")
    print(f"items scored: {b['n_items']} | disagreement set D: {b['n_disagree']} | cells: {b['cell_sizes']} | sharer own-template vs fuser-tokens agreement: {b['S_native_agreement']:.2f}")
    print(f"same-length partner found for the mismatched-sharer control on {100*(b['mismatch_partner_coverage_on_D'] or 0):.0f}% of D items")
    table([[k, v] for k, v in b["accuracy_all"].items() if k != "C_minus_R"] + [["C - R (paired)", b["accuracy_all"]["C_minus_R"]]], ["accuracy, all items", "[95% CI]"])
    print()
    table([[k, v] for k, v in b["accuracy_on_D"].items() if not k.endswith("_ci")], ["accuracy on D", "[95% CI]"])
    print("\nInheritance P(answer == sharer's answer). A: sharer right/receiver wrong | B: sharer WRONG/receiver right | W: both wrong, differ")
    rows = []
    for cn in ("A", "B", "W", "B+W"):
        r = b["inheritance"][cn]
        rows.append([cn, r["n"], fmt(r["Inh_C"], True), fmt(r["Rev_C"], True), fmt(r["Inh_C_mm"], True), fmt(r["Rev_C_mm"], True), fmt(r["E_excess"]), fmt(r["Inh_T"], True), fmt(r["Inh_T2"], True) if CFG["RUN_T2_RATIONALE"] else "-"])
    table(rows, ["cell", "n", "C2C inherits S", "C2C reverts to R", "null inherits S", "null reverts to R", "EXCESS E = C2C - null", "text handoff", "text+rationale"])
    print("(null = C2C with the sharer cache computed on a different same-length prompt. 'reverts' columns show the two perturbations are comparable in size.)")
    print(f"\nasymmetry Inh(C|A) - Inh(C|B): {fmt(b['asymmetry_Inh_C_A_minus_B'])}    bridge/text ratio in B: {b['bridge_over_text_B'] if b['bridge_over_text_B'] is None else round(b['bridge_over_text_B'], 2)}")
    print("predicted-letter marginals:", {k: {l: round(x, 2) for l, x in v.items()} for k, v in b["letter_marginals"].items()})
    print("mismatched-sharer letter leak (C_mm answer == partner's sharer answer):", b["mismatch_letter_leak_vs_partner_sharer_answer"])
    print("\nAUTO-READING (PREREG decision table):", b["reading"])
    print(f"non-finite logits so far: {COUNTERS['nonfinite']}")

def run_part_B():
    t_b = time.time()
    stage1(POOL); stage2(POOL); done = list(POOL)
    b = analyze_B([i["id"] for i in done])
    reserve = list(RESERVE)
    while reserve and (b["cell_sizes"]["A"] < CFG["MIN_CELL"] or b["cell_sizes"]["B"] < CFG["MIN_CELL"]):
        chunk, reserve = reserve[:400], reserve[400:]
        DEVIATIONS.append(f"extended Part B with {len(chunk)} reserve MMLU-Redux items because a cell had <{CFG['MIN_CELL']} items (PREREG allows this)")
        stage1(chunk); stage2(chunk); done += chunk
        b = analyze_B([i["id"] for i in done])
    b["seconds"] = round(time.time() - t_b)
    b["nonfinite_logits"] = COUNTERS["nonfinite"]
    return b, done

if CFG["RUN_B"]:
    RES["partB"], DONE_B = run_part_B()
    print_part_B(RES["partB"])
    try:
        RES["partB_run3"] = analyze_run3([i["id"] for i in DONE_B])
        print_run3(RES["partB_run3"])
    except Exception as e:   # records are complete; the analysis can be re-run offline from results.json
        import traceback; traceback.print_exc()
        DEVIATIONS.append(f"analyze_run3 raised {repr(e)[:200]}; records are saved, re-run the analysis offline")
```

```python
# --- 13. Part C: planted belief (sharer sees a false premise; receiver sees masked padding of equal length) ---
def build_plant(it, w):
    note = lambda L: f"Reference note: the correct answer to the question below is {L}.\n\n"
    plain = it["ids"]
    ids_w = encode_user(TOK_R, note(w) + it["prompt"])
    ids_x = encode_user(TOK_R, note("X") + it["prompt"])
    k = len(ids_w) - len(plain)
    if k <= 0 or len(ids_x) != len(ids_w):
        return None
    p = next((j for j in range(len(plain)) if plain[j] != ids_w[j]), len(plain))
    if ids_w[p + k:] != plain[p:] or ids_x[:p] != ids_w[:p] or ids_x[p + k:] != ids_w[p + k:]:
        return None
    n = len(ids_w) + len(RESP_IDS)
    mask_r = [1] * p + [0] * k + [1] * (n - p - k)
    ids_r = ids_w[:p] + [PAD_R] * k + ids_w[p + k:] + RESP_IDS     # receiver: premise replaced by masked padding
    return dict(ids_w=ids_w + RESP_IDS, ids_x=ids_x + RESP_IDS, ids_r=ids_r, mask_r=mask_r, L=len(ids_w), p=p, k=k)

def run_part_C():
    t_c = time.time()
    cand = [it for it in POOL if it["id"] in RECORDS and "R" in RECORDS[it["id"]]][: CFG["N_PLANT"] * 2]
    rows, skipped = [], 0
    for it in cand:
        if len(rows) >= CFG["N_PLANT"]:
            break
        w = random.Random(f"{SEED}:plant:{it['id']}").choice([l for l in LETTERS if l != it["y"]])
        b = build_plant(it, w)
        if b is None:
            skipped += 1
            continue
        c_plant = to_pred(c2c_logits(b["ids_r"], b["ids_w"], b["L"], mask_r=b["mask_r"]))
        c_ctrl = to_pred(c2c_logits(b["ids_r"], b["ids_x"], b["L"], mask_r=b["mask_r"]))
        s_plant, s_ctrl = to_pred(lm_logits(SHAR, b["ids_w"])), to_pred(lm_logits(SHAR, b["ids_x"]))
        t_plant = to_pred(lm_logits(RECV, b["ids_w"]))                                   # receiver is simply told, in text
        r_pad = to_pred(lm_logits(RECV, b["ids_r"], mask=b["mask_r"]))                  # what the receiver sees inside C_plant / C_ctrl, bridge off
        plain_full = it["ids"] + RESP_IDS                                                 # masking sanity: same tokens as plain, positions skip the masked gap,
        r_gap = to_pred(lm_logits(RECV, plain_full, positions=[j if j < b["p"] else j + b["k"] for j in range(len(plain_full))]))   # so R_pad must equal R_gap exactly (up to float noise)
        rows.append(dict(id=it["id"], y=it["y"], w=w, C_plant=c_plant, C_ctrl=c_ctrl, S_plant=s_plant, S_ctrl=s_ctrl,
                         T_plant=t_plant, R_pad=r_pad, R_gap=r_gap, R=RECORDS[it["id"]]["R"]))
        if len(rows) % 100 == 0:
            print(f"  part C {len(rows)}/{CFG['N_PLANT']}  {time.time()-t_c:.0f}s", flush=True)
    a = lambda k: np.array([r[k]["a"] for r in rows])
    w = np.array([r["w"] for r in rows])
    ok = np.ones(len(rows), bool)
    for k in ("C_plant", "C_ctrl", "S_plant", "S_ctrl", "T_plant", "R", "R_pad", "R_gap"):
        ok &= a(k) != "?"
    hit = lambda k: (a(k) == w).astype(float)
    susc = ok & (a("S_plant") == w) & (a("S_ctrl") != w)
    out = {"n": int(ok.sum()), "n_skipped_construction": skipped, "n_sharer_susceptible": int(susc.sum()),
           "P_follow": {k: fmt(boot_ci(hit(k)[ok]), True) for k in ("S_plant", "S_ctrl", "C_plant", "C_ctrl", "T_plant", "R")},
           "sharer_effect_S_plant_minus_S_ctrl": fmt(boot_ci((hit("S_plant") - hit("S_ctrl"))[ok])),
           "bridge_effect_all": fmt(boot_ci((hit("C_plant") - hit("C_ctrl"))[ok])),
           "bridge_effect_susceptible": fmt(boot_ci((hit("C_plant") - hit("C_ctrl"))[susc])),
           "bridge_effect_susceptible_ci": boot_ci((hit("C_plant") - hit("C_ctrl"))[susc]),
           "text_effect_T_plant_minus_R": fmt(boot_ci((hit("T_plant") - hit("R"))[ok])),
           "mask_sanity_same_answer": float((a("R_pad") == a("R_gap"))[ok].mean()) if ok.any() else None,
           "mask_sanity_mean_abs_dprob": float(np.mean([np.abs(np.array(r["R_pad"]["p"], float) - np.array(r["R_gap"]["p"], float)).mean() for r, k_ in zip(rows, ok) if k_])) if ok.any() else None,
           "receiver_pad_vs_plain_same_answer": float((a("R_pad") == a("R"))[ok].mean()) if ok.any() else None,
           "seconds": round(time.time() - t_c), "per_item": rows}
    lo = out["bridge_effect_susceptible_ci"][1]
    out["reading"] = ("BRIDGE CARRIES SHARER-PRIVATE INFORMATION (lower CI >= 0.05)" if lo >= 0.05 else
                      "no detectable transfer of the planted premise (exploratory; fuser was only trained on identical inputs)" if susc.sum() >= 30 else
                      "too few items where the sharer itself adopted the false premise; no claim")
    return out

def print_part_C(c):
    print("\n=== PART C: planted false belief (exploratory) ===")
    print(f"items: {c['n']} (skipped {c['n_skipped_construction']}) | sharer adopted the planted letter on {c['n_sharer_susceptible']}")
    table([[k, v] for k, v in c["P_follow"].items()], ["P(answer == planted letter w)", "[95% CI]"])
    print("sharer effect (S_plant - S_ctrl):      ", c["sharer_effect_S_plant_minus_S_ctrl"])
    print("bridge effect, all items:              ", c["bridge_effect_all"])
    print("bridge effect, sharer-susceptible items:", c["bridge_effect_susceptible"])
    print("text effect (receiver told directly):  ", c["text_effect_T_plant_minus_R"])
    print("mask sanity (masked-pad receiver vs plain receiver with the same position gap; should be ~identical):",
          "same answer", c["mask_sanity_same_answer"], "| mean |dprob|", c["mask_sanity_mean_abs_dprob"])
    print("   (receiver w/ masked pad vs plain receiver: same answer", c["receiver_pad_vs_plain_same_answer"], "; differs only by the positional gap)")
    print("READING:", c["reading"])

if CFG["RUN_C"] and CFG["RUN_B"]:
    # PREREG_run3 time-budget rule: Part C only if the time budget allows (minutes since notebook start + its estimate <= PART_C_WALL_MIN); otherwise dropped and logged
    _elapsed_min = (time.time() - T0) / 60
    if SMOKE or QUICK or _elapsed_min + est_C / 60 <= CFG["PART_C_WALL_MIN"]:
        RES["partC"] = run_part_C()
        print_part_C(RES["partC"])
    else:
        DEVIATIONS.append(f"Part C dropped (PREREG_run3 time-budget rule): {_elapsed_min:.0f} min elapsed + ~{est_C/60:.0f} min estimated > {CFG['PART_C_WALL_MIN']} min budget")
        print("!! Part C dropped:", DEVIATIONS[-1])
```

```python
# --- 14. Numeric task (PREREG_run3 safeguard 2): GSM8K exact-answer generation, no letters. R, S, C, C_mm; numeric inheritance + leak check ---
GSM8K_TEMPLATE = (
    "Solve the following math problem step by step. The last line of your response should be of the form Answer: $ANSWER (without quotes) where $ANSWER is the answer to the problem.\n\n"
    "{question}\n\n"
    "Please think step by step and explain your reasoning. Remember to put your answer on its own line after \"Answer:\", and you do not need to use a \\boxed command.")

def gsm8k_prompt(question):
    """Mirror of C2C UnifiedEvaluator._format_math_problem_example for gsm8k (checked against it by validate_notebook.py)."""
    return GSM8K_TEMPLATE.replace("{question}", question)

def gsm8k_gold(full):
    """Mirror of C2C UnifiedEvaluator.parse_answer for gsm8k."""
    if "####" in full:
        tail = full.split("####")[-1].strip()
        m = re.search(r"[-+]?\d+(?:\.\d+)?", tail)
        return m.group(0) if m else tail
    return None

def extract_number(text):
    """The number C2C's GSM8KEvaluator extracts from a solution (same math_verify call as GSMEvaluator.rule_judge); None if none."""
    try:
        ans = mv_parse(text, extraction_config=[LatexExtractionConfig(normalization_config=NormalizationConfig(nits=False, malformed_operators=False, basic_latex=True, boxed="all", units=True),
                                                                      boxed_match_priority=0, try_extract_without_anchor=False), ExprExtractionConfig()], extraction_mode="first_match")
    except Exception:
        return None
    for a in ans:
        try:
            return round(float(a), 6)
        except Exception:
            continue
    return None

def is_correct(text, gold):
    try:
        return bool(GSM.rule_judge(text, gold)[0])
    except Exception:
        return False

def numeq(a, b):
    return (a is not None) and (b is not None) and abs(a - b) <= 1e-6 * max(1.0, abs(a), abs(b))

def analyze_numeric(rows):
    """rows: dicts with gold, num{R,S,C,M} (float|None), ok{R,S,C,M} (bool), partner_S_num, partner_S_ok, ctrl_S_num.
    Decision input: verdict from the paired CI of acc(C) - acc(C_mm): positive = lower CI > 0, negative = upper CI < 0, null otherwise, unavailable if N < 60."""
    n = len(rows)
    if n < 2:
        return {"n": n, "verdict": "unavailable"}
    okv = {k: np.array([bool(r["ok"][k]) for r in rows], float) for k in "RSCM"}
    nm = {k: [r["num"][k] for r in rows] for k in "RSCM"}
    out = {"n": n, "acc": {k: fmt(boot_ci(okv[k]), True) for k in "RSCM"}, "acc_point": {k: float(okv[k].mean()) for k in "RSCM"},
           "no_number_extracted": {k: int(sum(x is None for x in nm[k])) for k in "RSCM"}}
    for a, b in (("C", "R"), ("C", "S"), ("C", "M"), ("M", "R")):
        out[f"{a}_minus_{b}_ci"] = boot_ci(okv[a] - okv[b])
        out[f"{a}_minus_{b}"] = fmt(out[f"{a}_minus_{b}_ci"])
    out["C_minus_Cmm_ci"] = out["C_minus_M_ci"]
    d = out["C_minus_M_ci"]
    out["verdict"] = "unavailable" if n < 60 else "positive" if d[1] > 0 else "negative" if d[2] < 0 else "null"
    eq = lambda a, b: np.array([numeq(x, y) for x, y in zip(a, b)], float)
    out["same_number_C_vs_Cmm"] = boot_ci(eq(nm["C"], nm["M"]))
    eCS, eMS = eq(nm["C"], nm["S"]), eq(nm["M"], nm["S"])
    haveS = np.array([x is not None for x in nm["S"]])
    Sw = haveS & (okv["S"] == 0)
    Rd = Sw & ~np.array([numeq(a, b) for a, b in zip(nm["R"], nm["S"])])
    out["inheritance"] = {name: {"n": int(m.sum()), "C_equals_S": boot_ci(eCS[m]), "C_mm_equals_S": boot_ci(eMS[m]), "excess": boot_ci((eCS - eMS)[m])}
                          for name, m in (("S_present", haveS), ("S_wrong", Sw), ("S_wrong_R_differs", Rd))}
    pS = [r["partner_S_num"] for r in rows]; cS = [r["ctrl_S_num"] for r in rows]
    leak, ctrl = eq(nm["M"], pS), eq(nm["M"], cS)
    has = np.array([a is not None and b is not None and c is not None for a, b, c in zip(nm["M"], pS, cS)])
    pw = has & np.array([not r["partner_S_ok"] for r in rows])
    out["leak"] = {name: {"n": int(m.sum()), "C_mm_equals_partner_S": boot_ci(leak[m]), "C_mm_equals_other_item_S": boot_ci(ctrl[m]), "excess": boot_ci((leak - ctrl)[m])}
                   for name, m in (("all", has), ("partner_S_wrong", pw))}
    return out

def build_numeric_groups():
    """GSM8K test items grouped by exact receiver-template token length. Within a group (>= 2 items, capped at 8) the partner of item j is item j+1
    (cyclic shift: no item is its own partner); partners are therefore also processed, so their sharer answers are known for the leak check."""
    ds = retry(lambda: load_dataset("openai/gsm8k", "main", split="test"))
    items = []
    for i, ex in enumerate(ds):
        gold = gsm8k_gold(ex["answer"])
        if gold is None:
            continue
        prompt = gsm8k_prompt(ex["question"])
        ids = encode_user(TOK_R, prompt)
        if len(ids) <= CFG["MAX_TOKENS"]:
            items.append(dict(id=f"gsm8k:{i}", prompt=prompt, ids=ids, L=len(ids), gold=gold))
    by_len = {}
    for it in items:
        by_len.setdefault(it["L"], []).append(it)
    glist = [sorted(g, key=lambda r: r["id"])[:8] for L, g in sorted(by_len.items()) if len(g) >= 2]
    random.Random(f"{SEED}:numgroups").shuffle(glist)
    groups, n = [], 0
    for g in glist:
        random.Random(f"{SEED}:numgrp:{g[0]['id']}").shuffle(g)
        for j, it in enumerate(g):
            it["partner"] = g[(j + 1) % len(g)]["id"]
        groups.append(g)
        n += len(g)
        if n >= CFG["N_NUM"]:
            break
    return groups, len(items)

def assemble_numeric_rows(raw):
    byid = {r["id"]: r for r in raw}
    rng = random.Random(f"{SEED}:numctrl")
    rows = []
    for r in raw:
        p = byid[r["partner"]]
        others = [x for x in raw if x["L"] != r["L"]]
        c = rng.choice(others) if others else None
        rows.append(dict(gold=r["gold"], num=r["num"], ok=r["ok"], partner_S_num=p["num"]["S"], partner_S_ok=p["ok"]["S"], ctrl_S_num=c["num"]["S"] if c else None))
    return rows

def run_numeric():
    t_n = time.time()
    groups, n_pool = build_numeric_groups()
    soft, hard = t_n + CFG["NUM_BUDGET_MIN"] * 60, T0 + CFG["NUM_WALL_MIN"] * 60
    print(f"numeric task: {n_pool} GSM8K test prompts <= {CFG['MAX_TOKENS']} tokens; {len(groups)} equal-length groups ({sum(len(g) for g in groups)} items) queued; "
          f"max_new_tokens {CFG['NUM_TOKENS']}; soft deadline {CFG['NUM_BUDGET_MIN']} min (once >= {CFG['N_NUM_MIN']} items), hard deadline at {CFG['NUM_WALL_MIN']} min since start", flush=True)
    NUM_ROWS.clear()
    stopped = None
    for gi, g in enumerate(groups):
        now = time.time()
        if now > hard or (now > soft and len(NUM_ROWS) >= CFG["N_NUM_MIN"]):
            stopped = f"stopped after {len(NUM_ROWS)} items ({'hard' if now > hard else 'soft'} deadline)"
            break
        idsl = [it["ids"] for it in g]
        txt = {}
        txt["R"] = hf_generate_batch(RECV, TOK_R, idsl, CFG["NUM_TOKENS"], EOS_R, PAD_R, bs=len(g))      # equal lengths: no padding, same numerics as unbatched
        txt["S"] = hf_generate_batch(SHAR, TOK_S, idsl, CFG["NUM_TOKENS"], EOS_S, PAD_S, bs=len(g))     # sharer fed the exact tokens the fuser feeds it
        txt["C"] = [c2c_generate(it["ids"], CFG["NUM_TOKENS"]) for it in g]
        byid = {it["id"]: it for it in g}
        txt["M"] = [c2c_generate(it["ids"], CFG["NUM_TOKENS"], ids_s=byid[it["partner"]]["ids"]) for it in g]   # sharer cache of a different same-length question
        for j, it in enumerate(g):
            tx = {k: txt[k][j].strip() for k in "RSCM"}
            NUM_ROWS.append(dict(id=it["id"], partner=it["partner"], L=it["L"], gold=it["gold"], text={k: v[:1500] for k, v in tx.items()},
                                 num={k: extract_number(v) for k, v in tx.items()}, ok={k: is_correct(v, it["gold"]) for k, v in tx.items()}))
        if gi == 0:
            print("  sample outputs (first item): " + " | ".join(f"{k}: {NUM_ROWS[0]['text'][k][-160:]!r}" for k in "RSCM"), flush=True)
        if (gi + 1) % 5 == 0:
            print(f"  numeric {len(NUM_ROWS)} items / {gi+1} groups  {time.time()-t_n:.0f}s", flush=True)
            save_partial()
    save_partial()
    rows = assemble_numeric_rows(NUM_ROWS)
    out = analyze_numeric(rows)
    out.update({"n_groups_done": gi + 1 if stopped is None else gi, "n_target": CFG["N_NUM"], "stopped": stopped, "max_new_tokens": CFG["NUM_TOKENS"], "seconds": round(time.time() - t_n), "per_item": NUM_ROWS})
    return out

def print_numeric(o):
    print("\n=== NUMERIC TASK: GSM8K exact-answer generation (greedy; C2C prompt and GSM8KEvaluator) ===")
    print(f"items: {o['n']} | groups: {o.get('n_groups_done')} | max_new_tokens {o.get('max_new_tokens')} | {o.get('stopped') or 'completed target'} | {o.get('seconds')}s")
    table([[k, o["acc"][k], o["no_number_extracted"][k]] for k in "RSCM"], ["condition (M = wrong-question cache)", "exact-match accuracy [95% CI]", "no number extracted"])
    print("paired:", {k: o[k] for k in ("C_minus_R", "C_minus_S", "C_minus_M", "M_minus_R")})
    print("same final number, C vs C_mm:", fmt(o["same_number_C_vs_Cmm"], True), "| VERDICT (decision input):", o["verdict"])
    table([[k, v["n"], fmt(v["C_equals_S"], True), fmt(v["C_mm_equals_S"], True), fmt(v["excess"])] for k, v in o["inheritance"].items()], ["numeric inheritance set", "n", "P(C num == S num)", "P(C_mm num == S num)", "excess"])
    table([[k, v["n"], fmt(v["C_mm_equals_partner_S"], True), fmt(v["C_mm_equals_other_item_S"], True), fmt(v["excess"])] for k, v in o["leak"].items()], ["leak check", "n", "P(C_mm num == partner's S num)", "P(C_mm num == random other item's S num)", "excess"])

if CFG["RUN_B"]:
    try:
        if GSM is None:
            raise RuntimeError(f"C2C GSM8K grader unavailable: {NUM_IMPORT_ERR}")
        RES["numeric"] = run_numeric()
        print_numeric(RES["numeric"])
        if RES["numeric"]["stopped"] or RES["numeric"]["n"] < CFG["N_NUM"]:
            DEVIATIONS.append(f"numeric task N={RES['numeric']['n']} < target {CFG['N_NUM']}: {RES['numeric']['stopped'] or 'queue exhausted'} (PREREG_run3 time-budget rule)")
    except Exception as e:   # never lose the MC results because of the numeric task
        import traceback; traceback.print_exc()
        DEVIATIONS.append(f"numeric task failed: {repr(e)[:250]}")
        RES["numeric"] = None
    try:
        RES["reading_final"], RES["reading_final_flags"] = auto_reading_run3(RES["partB_run3"], RES["numeric"]) if RES.get("partB_run3") else ("NO CLAIM: Part B analysis missing", {})
        print("\nFINAL READING (PREREG_run3 decision rule):", RES["reading_final"], RES["reading_final_flags"])
    except Exception as e:
        import traceback; traceback.print_exc()
        DEVIATIONS.append(f"final reading failed: {repr(e)[:200]}")
```
